```python
from mif import *
mif = load_mif()
print(mif)
mtxlist = [mif.gadget,mif.karateclub,mif.erdosReny,mif.scalefree,mif.homophilly,mif.heterophilly]
adjacencylist = [adjacencyinfocheck(i) for i in mtxlist]
```

```python
#Karate Club
#MiF value between the vertices 4 and 32 under the default settings.
i = 1
log = MiF(adjacencylist[i], 4, 32, 0.5, 3)
print(log)
```

```python
#homophily network setting the vertex 3 as the starting node.
#The default condition of loop = 0 doesn't allow the random walker to loop back to the same node.
#If loop = 1, there is no such constraint.
i = 4
log_without = MiF_broadcast(adjacencylist[i], 3)
log_with = MiF_broadcast(adjacencylist[i], 3, loop = 1)
```

```python
print(log_without)
print(log_with)
```

```python
#The pairwise comparison (difference) between the two broadcast outcomes (with or without self-loop) is checked by the function of MiF_broadcast_diff_on_loop.
MiF_broadcast_diff_on_loop(log_with, log_without)
```

```python
#The homophily network setting for the MiFDI computation the vertex with the least degree as the starting node (Default setting).
#The default condition of loop = 0 doesn't allow the random walker to loop back to the same node.
#If loop = 1, there is no such constraint.
logdi_without, dival_without = MiFDI(adjacencylist[i])
logdi_with, dival_with = MiFDI(adjacencylist[i],loop = 1)
```

```python
print(logdi_with)
print(logdi_without)
print(dival_with)
print(dival_without)
```

```python
MiFDI_diff_on_loop(logdi_with, logdi_without)
```

```python
#The homophily network setting for the MiFDI computation the vertex with the largest degree as the starting node (startingvertices = "max").
#The default condition of loop = 0 doesn't allow the random walker to loop back to the same node.
#If loop = 1, there is no such constraint.
logdi_max_without, dival_max_without = MiFDI(adjacencylist[i], startingvertices = "max")
logdi_max_with, dival_max_with = MiFDI(adjacencylist[i],startingvertices = "max",loop = 1)
```

```python
print(logdi_max_with)
print(logdi_max_without)
print(dival_max_with)
print(dival_max_without)
```

```python
MiFDI_diff_on_loop(logdi_max_with, logdi_max_without)
```
