```python
#Gadget network setting the vertex 3 as the starting node.
#The default condition of loop = 0 doesn't allow the random walker to loop back to the same node.
#If loop = 1, there is no such constraint.
i = 0
log_without0 = MiF_broadcast(adjacencylist[i], 3)
log_with0 = MiF_broadcast(adjacencylist[i], 3, loop = 1)
print(f"MiF brodacast without loop: {log_without0}")
print(f"MiF brodacast with loop: {log_with0}")
```

```python
#The pairwise comparison (difference) between the two broadcast outcomes (with or without self-loop) is checked by the function of MiF_broadcast_diff_on_loop.
MiF_broadcast_diff_on_loop(log_with0, log_without0)
```
